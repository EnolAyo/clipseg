```python
from third_party.Severstal.severstal_coco import DatasetCOCO
from torchvision import transforms
```

```python
data_path = '/home/eas/Enol/pycharm_projects/clipseg/third_party/Severstal/train_subimages'
mean = [0.34388125, 0.34388125, 0.34388125]
std = [0.13965334, 0.13965334, 0.13965334]
image_size = 256
transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])
split = 'train'
use_original_size = True
```

```python
coco = DatasetCOCO(data_path, transform, split, use_original_size)
```
